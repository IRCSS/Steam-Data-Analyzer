```python
from pathlib import Path
from PIL import Image

# Project paths
PROJECT_ROOT = Path.cwd()
IMAGES_DIR = PROJECT_ROOT / "images"
JPEG_DIR = IMAGES_DIR / "jpeg"

# Create output folder if needed
JPEG_DIR.mkdir(exist_ok=True)

png_files = sorted(IMAGES_DIR.glob("*.png"))

if not png_files:
    print("No PNG files found.")
else:
    for png_path in png_files:
        jpg_path = JPEG_DIR / f"{png_path.stem}.jpg"

        with Image.open(png_path) as img:
            # JPEG doesn't support transparency
            if img.mode in ("RGBA", "LA", "P"):
                background = Image.new("RGB", img.size, (255, 255, 255))
                background.paste(img, mask=img.split()[-1] if img.mode != "RGB" else None)
                img = background
            else:
                img = img.convert("RGB")

            img.save(jpg_path, "JPEG", quality=95)

        print(f"✓ {png_path.name} -> {jpg_path.name}")

print(f"\nDone! JPEGs saved to:\n{JPEG_DIR}")
```
